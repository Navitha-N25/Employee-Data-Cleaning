### Clean the dataset:

```python
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

print(df.columns.tolist())
```

```python
dtype_mapping = {
    "employee_id": "string",
    "age": "Int64",
    "salary": "string",
    "gender": "string",
    "department": "string",
    "position": "string",
    "joining_date": "string",
    "performance_score": "string",
    "email": "string",
    "phone_number": "string"
}
```

```python
dtype_mapping = {
    column: dtype
    for column, dtype in dtype_mapping.items()
    if column in df.columns
}
```

```python
for column in ["age"]:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column].replace("", pd.NA),
            errors="coerce"
        ).astype("Int64")
```

```python
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)
age_median = df["age"].median()

print("Median age:", age_median)
df["age"] = df["age"].fillna(
    round(age_median)
).astype("Int64")

print("Missing ages after cleaning:", df["age"].isna().sum())
```

### Convert date columns:

```python
date_columns = [
    column for column in df.columns
    if "date" in column.lower()
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

print(df[date_columns].dtypes)
```

### Standardize categorical columns:

```python
categorical_columns = [
    column for column in [
        "gender", "department", "position",
        "performance_score", "status", "city"
    ]
    if column in df.columns
]

for column in categorical_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
        .str.lower()
        .str.title()
    )

for column in categorical_columns:
    print(column, df[column].unique())
```

### Export the cleaned dataset:

```python
import os

folder_path = r"C:\Users\navih\OneDrive\Documents\Projects\MessyEmployee"

os.makedirs(folder_path, exist_ok=True)

print("Folder created successfully!")
```

```python
output_path = os.path.join(
    folder_path,
    "cleaned_employee_dataset.csv"
)
df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("File saved successfully:", output_path)
```
